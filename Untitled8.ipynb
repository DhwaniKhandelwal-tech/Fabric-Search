```python
import pandas as pd

# Replace 'your_file.csv' with the path to your CSV file
try:
    df = pd.read_csv('air_quality_historical.csv')
    print("CSV file loaded successfully!")
    # Display the first 5 rows of the DataFrame
    display(df.head())
except FileNotFoundError:
    print("Error: The file 'your_file.csv' was not found. Please check the file path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# Calculate mean of 'pm10' column
mean_pm10 = df['pm10'].mean()
print(f"Mean of pm10: {mean_pm10}")

# Calculate median of 'pm10' column
median_pm10 = df['pm10'].median()
print(f"Median of pm10: {median_pm10}")

# Calculate mode of 'pm10' column
mode_pm10 = df['pm10'].mode()
print(f"Mode of pm10:\n{mode_pm10}")
```
